```python
class playlist():
    def __init__(self, name):
        self.name = name
        self.songs = []
    def add_songs(self, song):
        self.songs.append(song)
        return self.songs
    def show_songs(self):
        for song in self.songs:
            print(song)
h1 = playlist('Ali')
h1.add_songs('Hey')
h1.add_songs('Hello')
h1.show_songs()
```

```python
class Parent():
    def __int__(self):
        print("Parent Initialized")
    def method(self):
        print("Parent Method")
class Child(Parent):
    def __init__(self):
        super().__init__()
        print("Child Initialized")
    def method(self):
        super().method()
        print("Child Method")
c2 = Child()
c2.method()

```

```python
class BankAccount():
    accounts = []
    def __init__(self, name, id,  initial_balance):
        self.name = name
        self.id = id
        self.__balance = initial_balance
        print(name, id, initial_balance)
    def Deposit(self, deposit):
        self.__balance += deposit
        print(self.name, self.id, self.__balance)
    def Withdraw(self, withdraw):
        self.withdraw = withdraw
        if withdraw <= self.__balance:
            self.__balance -= withdraw
        else:
            print("Insufficient Balance")
    def show_info(self):
        print(self.name, self.id, self.__balance)

class Account(BankAccount):
    def Deposit(self, deposit):
        user_type = int(input("Enter your type"))
        if user_type == 0:
            return super().Deposit(deposit)
        else:
            return "Not a user"
    def Withdraw(self, withdraw):
        user_type = int(input("Enter your type"))
        if user_type == 0:
            return super().Withdraw(withdraw)
        else:
            return "Not a user"
    def show_info(self):
        return super().show_info()
    
new = Account("ahmad",1,2000)
new.Deposit(100)
```

```python
class Dog():
    def __init__(self):
        pass
    def voice(self):
        print("Bark")
    def species(self):
        print("Pitbull")
class Cat():
    def __init__(self):
        pass
    def voice(self):
        print("Meow")
    def species(self):
        print("Felis catus")
class Horse():
    def __init__(self):
        pass
    def voice(self):
        print("neigh")
    def species(self):
        print("Equus caballus")
dog = Dog()
cat = Cat()
horse = Horse()
for x in (dog, cat, horse):
    x.voice()
    
```
